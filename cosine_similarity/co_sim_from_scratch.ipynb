```python
import numpy as np
#we create a function that we can reuse
def cosine_sim(vec_1, vec_2):  
    arr1 = np.array(vec_1)
    arr2 = np.array(vec_2)

    dot_prod = np.dot(arr1, arr2) # calculate the dot product 

    #this calculates the magnitude
    mg1 = np.linalg.norm(arr1)
    mg2 = np.linalg.norm(arr2)

    #this give use the final cosine similarity values
    sim = dot_prod / (mg1 * mg2)

    return sim 

a = [4,3,2]
b = [4,6,2]

print(cosine_sim(a,b))
```

```python
import torch
import torch.nn.functional as F

# Convert to PyTorch tensors (floats required)
a = torch.tensor([4.0, 3.0, 2.0])
b = torch.tensor([4.0, 6.0, 2.0])

# Dim=0 specifies computing similarity along the matching vector dimension
similarity = F.cosine_similarity(a, b, dim=0)

print(similarity.item())  # Output: 0.9192388653755188

```

```python
import torch
import torch.nn.functional as F

query = torch.tensor([4,3,2])

database = torch.tensor([
    [1.0, 0.0, 0.0],
    [4.0, 6.0, 2.0],
    [3.0, 3.0, 2.0],
    [0.0, 9.0, 1.0]
])

similarities = F.cosine_similarity(query, database, dim = 1)
best_match_idx = torch.argmax(similarities).item()

print("Similarity Scores:", similarities.tolist())
print(f"Closest Match is Document {best_match_idx} with a score of {similarities[best_match_idx].item():.4f}")
```
