```python
chunks = [
    "To reset your password, go to Settings > Security and click Reset.",
    "Error code E-404 means the server could not find the requested API endpoint.",
    "Refunds are processed within 5 to 7 business days after item return.",
    "If you encounter error code E-404 on the login page, clear your browser cache.",
    "Our customer support team is available Monday through Friday from 9 AM to 5 PM."
]
```

```python
def tokenize(text):
    return re.findall(f'\w+', text.lower())

tokenized_chunks = [tokenize(chunk) for chunk in chunks]

bm25 = BM25Okapi(tokenized_chunks)
query = "How do I fix error E-404?"
tokenized_query = tokenize(query)
scores = bm25.get_scores(tokenized_query)

results = sorted(zip(chunks,scores), key = lambda x: x[1], reverse=True)

print(f"Query: '{query}'\n")
for chunk, score in results[:4]:  # Print top 3 chunks
    print(f"Score: {score:.4f} | Chunk: {chunk}")
```

```python
bm25_ = BM25Okapi(tokenized_chunks)
query_ = 'after how many days are refunds processed'

tokenized_query_ = tokenize(query_)

scores_ = bm25_.get_scores(tokenized_query_)

results_ = sorted(zip(chunks,scores_), key = lambda x: x[1], reverse=True)
print(f"Query: {query_}")
for chunk, score in results_[:4]:
    print(f"Score: {score:.4f} | Chuck: {chunk}")
```
